🚀 PROJETO: Análise de Vendas (Pipeline de Dados)


🎯 Objetivo

Analisar dados de vendas para identificar:

- produtos mais vendidos
- faturamento por região
- faturamento por mês
- percentual de cresciemento (mês a mês)
- ranking por cidade
- comportamento ao longo do tempo


📊 1. Dataset - Descrição da Base de Dados — Vendas

Esta base de dados contém registros de vendas realizadas em diferentes regiões do Brasil, abrangendo informações de clientes, produtos, categorias, valores, quantidade e cidade.



🧠 2. Código completo (pipeline + análise)

```python
# 1 - Importação de bibliotecas
import pandas as pd
```

```python
# 2- Extração dos dados

df = pd.read_csv('../dados/vendas.csv')

```

```python
# 3 - Exploração
print(df.head())
print(df.info)
```

```python
# 4 - Limpeza dos dados
df = df.dropna()  # Remove linhas com valores ausentes
df = df.drop_duplicates()  # Remove linhas duplicadas
```

```python
# Converter data
df['data'] = pd.to_datetime(df['data'], errors='coerce')
```

```python
# Criar coluna total
df['total'] = df['valor'] * df['quantidade']
```

```python
# 5 - Análise

#faturamento total
total_vendas = df['total'].sum()
print(f'Total de vendas: {total_vendas}')
```

```python
# Por produto
vendas_por_produto = df.groupby('produto')['total'].sum()
print('Vendas por produto:')
print(vendas_por_produto)
```

```python
# Por cidade
vendas_por_cidade = df.groupby('cidade')['total'].sum()
print('Vendas por cidade:')
print(vendas_por_cidade)
```

```python
# Por mês
df['mes'] = df['data'].dt.month
vendas_por_mes = df.groupby('mes')['total'].sum()
print('Vendas por mês:')
print(vendas_por_mes)
```

📈 Crescimento mensal

```python
mensal = df.groupby('mes')['total'].sum()
crescimento = mensal.pct_change() * 100

print('Crescimento percentual mês a mês:')
print(crescimento.map("{:.2f}%".format))
```

🏆 Top produtos

```python
top_produtos = df.groupby('produto')['total'].sum().sort_values(ascending=False)
print('Top 5 produtos mais vendidos:')
print(top_produtos.head(5))
```

🌎 Ranking por cidade

```python
ranking_cidade = df.groupby('cidade')['total'].sum().sort_values(ascending=False)
print('Ranking de cidades por vendas:')
print(ranking_cidade)
```

```python
# 6 - Exportação dos resultados
vendas_por_produto.to_csv('../dados/resultados_vendas_por_produto.csv')
vendas_por_cidade.to_csv('../dados/resultados_vendas_por_cidade.csv')
vendas_por_mes.to_csv('../dados/resultados_vendas_por_mes.csv')
top_produtos.to_csv('../dados/resultados_top_produtos.csv')
ranking_cidade.to_csv('../dados/resultados_ranking_cidade.csv')
```
